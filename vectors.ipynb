**Scalar multipication of a vector **

```python
v = [2,4]
scalar = 2
result = (scalar*v[0] ,scalar*v[1])
print(result)
```

```python
import numpy as np
v = np.array([2,4])
scalar= 2
result = scalar *v
print(result)
```

Addition/subtraction → two vectors are involved, and the magnitude/direction of the result can change.

Scalar multiplication → one vector is involved, and the scalar scales its magnitude (and a negative scalar reverses its direction).

```python
v =[3,4]
magnitude = (v[0]**2 +v[1]**2)**0.5
print(magnitude)

```

```python
import numpy as np
v = np.array([3,4])
magnitude = np.linalg.norm(v)
print(magnitude)
```
